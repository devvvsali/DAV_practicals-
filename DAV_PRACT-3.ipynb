step 1 : import library

```python
# data set

import pandas as pd
import numpy as np
```

 create a dataset and dataframe


```python

data = {
    "Emp_id":[101,102,103,104,105,106,107,108],
    "Name":["Rahul","Priya","Amit","Riya","karan","Neha","Vivek","Pooja"],
    "Department":["IT","HR","IT","Finance","HR","Finance","IT","HR"],
    "City":["Surat","Navsari","Surat","Valsad","Surat","Navsari","Valsad","Surat"],
    "Salary":[45000,38000,52000,48000,40000,51000,60000,42000],
    "Experience":[2,1,5,4,2,6,8,3]
}
```

```python
df=pd.DataFrame(data)
df
```

step 2: filter the data

step 2.1: display the employee  from IT department

```python
# Step 2 : Filter Data
df[df["Department"]=="IT"]
```

Step 2.2:display employees salary over 50000

```python
# dipaly salary over 50000
df[df["Salary"]>50000]
```

Step 2.3 display employee from surat

```python
#display employess from surat city
df[df["City"]=="Surat"]
```

Step 2.4: display employee from IT deparment haviing salary more than 50000

```python
#display employee from IT deparment haviing salary more than 50000
df[(df["Department"]=="IT")&(df["Salary"]>50000)]
```

step 3: sorting data

Step 3.1:sorting salary in ascending order

```python
#ascending order
df.sort_values(by="Salary")
```

Step 3.2:sorting salary in descending order

```python
#descending order
df.sort_values(by="Salary", ascending=False)
```

Step 4:perfrom aggregation

Step 4.1: Calculate total salary of employee

```python
#employee salary total
df["Salary"].sum()
```

Step 4.2: calculate the avg salary

```python
#avg salary
df["Salary"].mean()
```

Step 4.3: find the highest salary

```python
#highest salary
df["Salary"].max()
```

Step 4.4: find the lowest salary

```python
# lowest salary
df["Salary"].min()
```

Step 5 : perfrom Group by operations

step 5.1 group employees based on department& calculate avg salary of each department

```python
# Group by employee and calculate avg salary based on department
df.groupby("Department")["Salary"].mean()
```

5.2 calculate the total slary paid by each deparment

```python
#Total salary
df.groupby("Department") ["Salary"].sum()
```

```python
# display the highest salary in each deparment
df.groupby("Department")["Salary"].max()
```

```python
#calculate the no of employee in each department
df.groupby("Department")["Salary"].count()
```

Step 6: multiple  Aggregation function

```python
#using multiple aggregation
df.groupby("Department")["Salary"].agg(["mean","sum","max","min","count"])
```
