step 1: import pandas library


```python
#assign the  pandas as pd
import pandas as pd
```

step 2:create sample data set


```python
# data we created

data={

       "ID": [ 101, 102, 103, 104, 105, 106, 107],
       "Name" :["Rahul", "Priya", "Amit", "Riya", "Karan","krisha", " Harry"],
       "Age": [20, 21, 19, 20 , 22, 24, 28 ],
       "Marks": [100, 91, 90, 28 , 82, 94, 60 ],
       "City" : [ "Surat", "Navsari", "Valsad","Bharuch","Mumbai", "Banglore","pune"]

}
```

step 3: create a data frame

```python
# converting the dictionary into pandas dataframe
df = pd.DataFrame(data)
print(df)
```

step 4: save data as an csv file


```python
#convert the data in csv
df.to_csv("students.csv", index=False)
#display the message
print("csv file is created ")
```

step 5: save data as an excel file


```python
#convert the data in excel
df.to_excel("students.xlsx", index=False)
#display the message
print("excel file is created")
```

step 6: save data as an json file


```python
#convert the data in json
df.to_json("students.json", index=False)
#display the message
print("json file is created")
```
