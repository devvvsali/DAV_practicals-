Step - 1  Import Required Libraries

```python
#Import the Pandas LIbrary
# Pandas is used
import pandas as pd
#import the matplotlib
import matplotlib.pyplot as plt
```

step-2 create a data set

```python
# creating  a data set
data={

       "Month": [ "January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"],
       "Unit_Sold" :[120, 135, 150, 145, 160, 175, 180, 170, 190, 210,  230, 250],
       "Revenue" : [ 120000, 135000, 150000, 145000, 160000, 175000, 180000, 170000, 190000, 210000, 230000, 250000],
       "Profit" :[20000, 23000, 25000, 24000, 27000, 30000, 31000, 29000, 33000, 37000, 41000, 45000]

}
df = pd.DataFrame(data)
print(df)

```

step: 3 Inspect the dataset

```python
df.head()# display first five
```

```python
# display ythe info
df.info()
```

Step: 4 Generate discriptive statistic

```python
#
df.describe()
```

step 5 calculate mean

```python

# calculate the avg numbers of uint for permonth

average_units = df['Unit_Sold'].mean()

print("Average units sold per month:", average_units)
```

Step: 6 calculate median

```python
# calcutae the middel value of the units_Sold column
median_units = df['Unit_Sold'].median()

print("Median units sold per month:", median_units)
```

Step: 7 calculate max and min

```python

#caluate ther max and mini
# find the highest no of units sold in month
maximum_units = df['Unit_Sold'].max()
minimum_units = df['Unit_Sold'].min()

print("Maximum units sold per month:", maximum_units)
print("Minimum units sold per month:", minimum_units)
```

Step: 8 calculate standard deviation

```python

#calculate the standar devaition unit sold
std_units = df["Unit_Sold"].std()
print("Standard deviation of units sold per month:", std_units)
```

Step: 9 find the month with highest sales

```python
# find the month with highest  sales
# find the index of the row  having the maximum revenue.
highest_sales_index = df["Revenue"].idxmax()

# retive the month correspondig to the index
highest_sales_month = df.loc[highest_sales_index, "Month"]
print("Month with highest sales:", highest_sales_month)
# retrive the maximum revenue
highest_revenue = df.loc[highest_sales_index, "Revenue"]
print("Highest revenue:", highest_revenue)
```

Step: 10 find the month with lowest sales

```python
#  find the ,month with lowest sales

lowest_sales_index = df["Revenue"].idxmin()
#retirve the month correspinding
lowest_sales_month = df.loc[lowest_sales_index, "Month"]
print("Month with lowest sales:", lowest_sales_month)
# lowest revenue
lowest_revenue = df.loc[lowest_sales_index, "Revenue"]
print("Lowest revenue:", lowest_revenue)
```

Step : 11 Calculate the total revenue

```python

# calculate the lowest revenue generated during this year
total_revenue = df["Revenue"].sum()
print("Total revenue:", total_revenue)
```

step : 12 calculate total profit

```python
# calculate total profit
# Calculate the total profit generated during the year
total_profit = df["Profit"].sum()
print("Total profit:", total_profit)
```

Step:13 identify the sales

```python

# identify the sales Trend
# create a line chart  to visualize monthly revneue
plt.figure(figsize=(10, 5))
# ploting the revenue graph
plt.plot(df["Month"],df["Revenue"], marker='o', linestyle='-')
# LABELING X AXIS
plt.xlabel("Month")
# LABELING Y AXIS
plt.ylabel("Revenue")
# add the titel on the graph.
plt.title("Monthly Revenue")
# rotate the months name in the graph
plt.xticks(rotation=45)
# show the graph
plt.show()
```

Step:14 Analyze profit

```python
# create a line chart for monthly profit
plt.figure(figsize=(10, 5))
# PLOT TH GRAPH PROFIT
plt.plot(df["Month"],df["Profit"], marker='o', linestyle='--')
# LABEL FOR X AXIS
plt.xlabel("Month")
# LABEL FOR Y AXIS
plt.ylabel("Profit")
# TITLE FOR THE GRAPH
plt.title("Monthly Profit")
plt.xticks(rotation=45)
plt.show()
```

Step:15 compare revenue and profit both

```python
#  comparee revenve and profit
plt.figure(figsize=(10, 5))
# plot revenue
plt.plot(df["Month"], df["Revenue"], marker='o', linestyle='--', label='Revenue')
# plot profit
plt.plot(df["Month"], df["Profit"], marker='o', linestyle='--', label='Profit')
# label for x - axis label
plt.xlabel("Month")
# label for y axis label
plt.ylabel("Amount")
# title for the graphs plot
plt.title("Revenue vs Profit")
# rotate the months naeme for better reading
plt.xticks(rotation=45)
# display the two lines
plt.legend()
# to print the graph
plt.show()
```

Step:16 Identify growth in sales

```python

# calculate the percentage change in revenue
# from one month to the next
df["Revenue_Growth_%"] = df["Revenue"].pct_change()*100
# display the data set
df
```

Step:17 Identify the months with increasing revenue

```python
# identify the months with increasing Revenue
# Filter the dataset to display only those monthns
# where the growth is postive
increasing_revenue_months = df[df["Revenue_Growth_%"] > 0]
# display the months
increasing_revenue_months
```
